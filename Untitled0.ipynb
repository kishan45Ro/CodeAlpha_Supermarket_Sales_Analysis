```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))

sns.barplot(
    x='Product line', y='Sales', data=df,
    estimator=sum, errorbar=None, palette='viridis'
)

plt.title('Total Sales by Product Line', fontsize=16)
plt.xlabel('Product Line', fontsize=12)
plt.ylabel('Total Sales', fontsize=12)
plt.xticks(rotation=45)
plt.show()
```

```python

gender_sales = df.groupby('Gender')['Sales'].sum()

plt.figure(figsize=(6, 6))
plt.pie(gender_sales, labels=gender_sales.index, autopct='%1.1f%%', colors=['#ff9999','#66b3ff'], startangle=90)
plt.title('Sales Distribution by Gender', fontsize=16)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(x='City', data=df, palette='Set2')
plt.title('Number of Transactions by City', fontsize=16)
plt.xlabel('City', fontsize=12)
plt.ylabel('Number of Bills', fontsize=12)
plt.show()
```
